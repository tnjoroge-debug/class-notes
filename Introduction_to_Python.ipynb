### 5. Object-Oriented

Python supports **object-oriented programming (OOP)**.

This means Python allows programmers to create and work with:

* **Classes** – blueprints for creating objects
* **Objects** – instances created from classes

* **Methods** – functions that belong to objects

For example:

```python

```

Object-oriented programming helps programmers organize large and complex programs into reusable components.

# 30. Operator Precedence

When an expression contains several operators, Python follows an order of evaluation.

For example:

```python
result = 10 + 5 * 2
```

Multiplication occurs first.

Therefore:

```text
10 + 10 = 20
```

Parentheses can change the order:

```python
result = (10 + 5) * 2
```

Result:

```text
30
```

### General order

1. Parentheses
2. Exponents
3. Multiplication/division
4. Addition/subtraction

# 51. Scope of Variables

Beginners should generally avoid unnecessary global variables.

# 56. Common Beginner Errors

### 1. Incorrect indentation

Incorrect:

```python
if age >= 18:
print("Adult")
```

Correct:

```python
if age >= 18:
    print("Adult")
```

### 2. Using `=` instead of `==`

Assignment:

```python
age = 20
```

Comparison:

```python
age == 20
```

### 3. Forgetting quotation marks

Incorrect:

```python
name = John
```

Correct:

```python
name = "John"
```

### 4. Mixing strings and numbers

This can cause an error:

```python
age = 20
print("Age: " + age)
```

Better:

```python
print("Age:", age)
```

or:

```python
print(f"Age: {age}")
```

### 5. Incorrect variable names

Python is case-sensitive:

```python
name = "John"

print(Name)
```

This causes an error because `name` and `Name` are different identifiers.
