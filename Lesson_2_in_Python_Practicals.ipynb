# Student Management Program-Code

Build a **simple student management program** that:

1. Asks for a student's name.
2. Accepts three marks.
3. Calculates the average.
4. Determines the grade.
5. Displays the student's result.
6. Uses at least one function.
7. Handles invalid numerical input.


```python
# ============================================================
#              STUDENT MANAGEMENT PROGRAM
# ============================================================

# FUNCTION 1: CALCULATE GRADE
def calculate_grade(average):

    if average >= 70:
        return "A"

    elif average >= 60:
        return "B"

    elif average >= 50:
        return "C"

    elif average >= 40:
        return "D"

    else:
        return "F"


# FUNCTION 2: GET VALID MARK
def get_mark(subject):

    while True:

        try:
            mark = float(input(f"Enter marks for {subject} (0-100): "))

            if 0 <= mark <= 100:
                return mark

            else:
                print("Invalid mark!")
                print("Please enter a mark between 0 and 100.")

        except ValueError:
            print("Invalid input!")
            print("Please enter a numerical value.")


# ============================================================
# MAIN PROGRAM
# ============================================================

# Ask for student's name
student_name = input("Enter student's name: ")


# Ask for three marks
mark1 = get_mark("Subject 1")
mark2 = get_mark("Subject 2")
mark3 = get_mark("Subject 3")


# Calculate total marks
total = mark1 + mark2 + mark3


# Calculate average
average = total / 3


# Determine grade
grade = calculate_grade(average)


# Determine performance
if grade == "A":
    performance = "Excellent"

elif grade == "B":
    performance = "Very Good"

elif grade == "C":
    performance = "Good"

elif grade == "D":
    performance = "Pass"

else:
    performance = "Fail"


# ============================================================
# DISPLAY STUDENT RESULT
# ============================================================

print()
print("=" * 50)
print("             STUDENT RESULT")
print("=" * 50)

print(f"Student Name : {student_name}")

print("-" * 50)

print(f"Subject 1    : {mark1:.2f}")
print(f"Subject 2    : {mark2:.2f}")
print(f"Subject 3    : {mark3:.2f}")

print("-" * 50)

print(f"Total Marks  : {total:.2f}")
print(f"Average      : {average:.2f}")
print(f"Grade        : {grade}")
print(f"Performance  : {performance}")

print("=" * 50)
```

# Example 2: Simple Shopping/Billing Program

This example teaches:

* Variables
* User input
* `float()` and `int()`
* Functions
* Calculations
* `if/elif/else`
* Error handling with `try/except`
* Formatted output


```python
# ============================================================
#              SIMPLE SHOPPING BILLING PROGRAM
# ============================================================

# FUNCTION 1: GET VALID NUMBER
def get_number(message):
    """
    This function asks the user to enter a number.
    It prevents the program from crashing when
    the user enters invalid numerical input.
    """

    while True:

        try:
            number = float(input(message))

            if number >= 0:
                return number

            else:
                print("Please enter a number that is 0 or greater.")

        except ValueError:
            print("Invalid input!")
            print("Please enter a numerical value.")


# FUNCTION 2: CALCULATE DISCOUNT
def calculate_discount(total):
    """
    This function calculates the discount based
    on the total shopping amount.

    10% discount for totals of 5,000 or more
    5% discount for totals of 2,000 to 4,999
    No discount for totals below 2,000
    """

    if total >= 5000:
        discount_rate = 0.10

    elif total >= 2000:
        discount_rate = 0.05

    else:
        discount_rate = 0

    discount_amount = total * discount_rate

    return discount_amount


# ============================================================
# MAIN PROGRAM
# ============================================================

print("=" * 55)
print("             SIMPLE SHOPPING BILLING SYSTEM")
print("=" * 55)

# Ask for customer's name
customer_name = input("Enter customer's name: ")

print()
print("Enter the details of the items purchased.")
print()


# ------------------------------------------------------------
# ITEM 1
# ------------------------------------------------------------

print("ITEM 1")

item1 = input("Enter item name: ")
price1 = get_number("Enter price: KSh ")
quantity1 = get_number("Enter quantity: ")


# ------------------------------------------------------------
# ITEM 2
# ------------------------------------------------------------

print()
print("ITEM 2")

item2 = input("Enter item name: ")
price2 = get_number("Enter price: KSh ")
quantity2 = get_number("Enter quantity: ")


# ------------------------------------------------------------
# ITEM 3
# ------------------------------------------------------------

print()
print("ITEM 3")

item3 = input("Enter item name: ")
price3 = get_number("Enter price: KSh ")
quantity3 = get_number("Enter quantity: ")


# ============================================================
# CALCULATE COST OF EACH ITEM
# ============================================================

cost1 = price1 * quantity1
cost2 = price2 * quantity2
cost3 = price3 * quantity3


# Calculate total before discount
subtotal = cost1 + cost2 + cost3


# Calculate discount
discount = calculate_discount(subtotal)


# Calculate amount after discount
amount_after_discount = subtotal - discount


# Calculate tax
tax = amount_after_discount * 0.16


# Calculate final amount
final_amount = amount_after_discount + tax


# ============================================================
# DISPLAY CUSTOMER BILL
# ============================================================

print()
print()
print("=" * 65)
print("                     CUSTOMER BILL")
print("=" * 65)

print(f"Customer Name: {customer_name}")

print("-" * 65)

print(f"{'Item':<20}{'Price':>12}{'Quantity':>12}{'Cost':>15}")
print("-" * 65)

print(f"{item1:<20}{price1:>12.2f}{quantity1:>12.0f}{cost1:>15.2f}")

print(f"{item2:<20}{price2:>12.2f}{quantity2:>12.0f}{cost2:>15.2f}")

print(f"{item3:<20}{price3:>12.2f}{quantity3:>12.0f}{cost3:>15.2f}")

print("-" * 65)

print(f"{'Subtotal:':<45} KSh {subtotal:>10.2f}")
print(f"{'Discount:':<45} KSh {discount:>10.2f}")
print(f"{'Amount After Discount:':<45} KSh {amount_after_discount:>10.2f}")
print(f"{'VAT (16%):':<45} KSh {tax:>10.2f}")

print("-" * 65)

print(f"{'FINAL AMOUNT:':<45} KSh {final_amount:>10.2f}")

print("=" * 65)

print()
print("Thank you for shopping with us!")
```

This example teaches:

## Problem Description

Build a **simple ATM/Banking System** that allows a user to perform basic banking operations.

The program should:

1. Ask the user to enter their name.
2. Start with an initial account balance.
3. Display an ATM menu.
4. Allow the user to:

   * Check account balance
   * Deposit money
   * Withdraw money
   * Exit the system
5. Validate numerical input.
6. Prevent the user from withdrawing more money than the available balance.
7. Use functions to organize the program.
8. Continue displaying the menu until the user chooses to exit.
9. Display appropriate messages for invalid choices.
10. Display the final account balance when the user exits.

## Suggested ATM Menu

The program should display a menu similar to:

**ATM MENU**

1. Check Balance
2. Deposit Money
3. Withdraw Money
4. Exit

The user should select an option and the program should perform the corresponding operation.

## Expected Learning Outcome

After completing this project, students should be able to combine **variables, input, functions, conditions, loops, calculations, and error handling** to build a simple real-world Python application.


```python
# ============================================================
#              EXAMPLE 3: SIMPLE ATM/BANKING SYSTEM
# ============================================================

# ------------------------------------------------------------
# FUNCTION 1: DISPLAY ATM MENU
# ------------------------------------------------------------

def display_menu():
    print()
    print("=" * 45)
    print("                 ATM MENU")
    print("=" * 45)
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Exit")
    print("=" * 45)


# ------------------------------------------------------------
# FUNCTION 2: GET NUMERICAL INPUT
# ------------------------------------------------------------

def get_amount(message):

    while True:

        try:
            amount = float(input(message))

            if amount > 0:
                return amount

            else:
                print("Invalid amount!")
                print("Amount must be greater than zero.")

        except ValueError:
            print("Invalid input!")
            print("Please enter a numerical value.")


# ------------------------------------------------------------
# FUNCTION 3: CHECK BALANCE
# ------------------------------------------------------------

def check_balance(balance):

    print()
    print("-" * 45)
    print(f"Current Balance: KSh {balance:,.2f}")
    print("-" * 45)


# ------------------------------------------------------------
# FUNCTION 4: DEPOSIT MONEY
# ------------------------------------------------------------

def deposit_money(balance):

    amount = get_amount("Enter amount to deposit: KSh ")

    balance = balance + amount

    print()
    print(f"Deposit successful!")
    print(f"Amount deposited: KSh {amount:,.2f}")
    print(f"New balance: KSh {balance:,.2f}")

    return balance


# ------------------------------------------------------------
# FUNCTION 5: WITHDRAW MONEY
# ------------------------------------------------------------

def withdraw_money(balance):

    amount = get_amount("Enter amount to withdraw: KSh ")

    if amount > balance:

        print()
        print("Withdrawal failed!")
        print("Insufficient funds.")
        print(f"Available balance: KSh {balance:,.2f}")

    else:

        balance = balance - amount

        print()
        print("Withdrawal successful!")
        print(f"Amount withdrawn: KSh {amount:,.2f}")
        print(f"Remaining balance: KSh {balance:,.2f}")

    return balance


# ============================================================
# MAIN PROGRAM
# ============================================================

print("=" * 55)
print("             WELCOME TO PYTHON BANK")
print("=" * 55)


# ------------------------------------------------------------
# STEP 1: GET CUSTOMER NAME
# ------------------------------------------------------------

customer_name = input("Enter your name: ")


# ------------------------------------------------------------
# STEP 2: SET INITIAL ACCOUNT BALANCE
# ------------------------------------------------------------

balance = 10000.00


print()
print(f"Welcome, {customer_name}!")
print(f"Your opening balance is: KSh {balance:,.2f}")


# ------------------------------------------------------------
# STEP 3: DISPLAY MENU REPEATEDLY
# ------------------------------------------------------------

while True:

    display_menu()

    # Get the user's menu choice
    try:
        choice = int(input("Enter your choice (1-4): "))

    except ValueError:
        print()
        print("Invalid input!")
        print("Please enter a number from 1 to 4.")
        continue


    # --------------------------------------------------------
    # OPTION 1: CHECK BALANCE
    # --------------------------------------------------------

    if choice == 1:

        check_balance(balance)


    # --------------------------------------------------------
    # OPTION 2: DEPOSIT MONEY
    # --------------------------------------------------------

    elif choice == 2:

        balance = deposit_money(balance)


    # --------------------------------------------------------
    # OPTION 3: WITHDRAW MONEY
    # --------------------------------------------------------

    elif choice == 3:

        balance = withdraw_money(balance)


    # --------------------------------------------------------
    # OPTION 4: EXIT
    # --------------------------------------------------------

    elif choice == 4:

        print()
        print("=" * 55)
        print("              ACCOUNT SUMMARY")
        print("=" * 55)

        print(f"Customer Name : {customer_name}")
        print(f"Final Balance : KSh {balance:,.2f}")

        print("=" * 55)
        print("Thank you for using Python Bank!")
        print("Goodbye!")
        print("=" * 55)

        # Stop the while loop
        break


    # --------------------------------------------------------
    # INVALID MENU CHOICE
    # --------------------------------------------------------

    else:

        print()
        print("Invalid choice!")
        print("Please select an option from 1 to 4.")
```
